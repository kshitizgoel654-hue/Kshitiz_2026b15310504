## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
# Keep the CSV in the same folder as your notebook, or change this path.
df = pd.read_csv("Student_09_dataset.csv")

# Display the first five records.
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Shape (rows, columns):", df.shape)
print("Column names:", list(df.columns))

print("\nData types and non-missing counts:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

# 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# Select the four evaluation-score columns and convert them to a NumPy array.
score_columns = ["Quality", "Innovation", "Presentation", "Teamwork"]
scores = df[score_columns].to_numpy()

# Calculate the overall average and the highest and lowest individual scores.
print("Average evaluation score:", round(np.mean(scores), 2))
print("Highest individual score:", np.max(scores))
print("Lowest individual score:", np.min(scores))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Add each student's average across the four criteria.
df["Average_Score"] = df[score_columns].mean(axis=1).round(2)

# Add a weighted score using the assignment's percentages.
df["Weighted_Score"] = (
    df["Quality"] * 0.30
    + df["Innovation"] * 0.25
    + df["Presentation"] * 0.25
    + df["Teamwork"] * 0.20
).round(2)

# Filter to show records with a weighted score of at least 80.
high_scoring = df[df["Weighted_Score"] >= 80]
print("Students with a weighted score of at least 80:")
display(high_scoring[["Student_ID", "Name", "Weighted_Score"]])
```

## 6. Your individual project challenge
**Challenge:** Calculate weighted score (30%, 25%, 25%, 20%); rank projects and identify top three.

Write your solution below and explain what the result means.


```python
# Assign rank 1 to the highest weighted score.
df["Rank"] = df["Weighted_Score"].rank(
    method="min",
    ascending=False
).astype(int)

# Sort from highest score to lowest.
ranked_projects = df.sort_values(
    by=["Rank", "Student_ID"]
)

print("Highest and lowest weighted scores:")
display(ranked_projects.iloc[[0, -1]][
    ["Rank", "Student_ID", "Name", "Weighted_Score"]
])

print("Top three students:")
display(ranked_projects.head(3)[
    ["Rank", "Student_ID", "Name", "Weighted_Score"]
])
```

## 7. Export your report
Use `to_csv(..., index=False)` to save your cleaned or analysed DataFrame. Create an `output` folder if needed.


```python
# Save the ranked analysis as a CSV file in the notebook's folder.
ranked_projects.to_csv("Student_09_Final_Analysis.csv", index=False)

print("Saved Student_09_Final_Analysis.csv")
```
